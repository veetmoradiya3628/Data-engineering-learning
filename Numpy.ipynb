```python
array1 = np.array([1, 2, 3, 4, 5])
array2 = np.array([6, 7, 8, 9, 10])

print(array1 + array2)
print(array1 - array2)
print(array1 * array2)
print(array1 / array2)
```

```python
print(np.max(array1))
print(np.min(array1))
print(np.max(array2))
print(np.min(array2))
```

```python
greater_than = array1 > 3
print(greater_than)
```

```python
result = array1 + 10
print(result)
```

```python
arr1 = np.array([10, 20, 30, 40, 50])

print(arr1[1])

print(arr1[-2])
```

```python
slice_arr = arr1[1:4]
print(slice_arr)
```

```python
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

mul = np.dot(arr1, arr2)
print(mul)
```

```python
invr = np.linalg.inv(arr1)
print(invr)
```

```python
eval, evect = np.linalg.eig(arr1)
print(eval)
print(evect)
```

```python
data = np.array([1, 2, 3, 4, 5])
mean = np.mean(data)
std_dev = np.std(data)
print(mean)
print(std_dev)
```
