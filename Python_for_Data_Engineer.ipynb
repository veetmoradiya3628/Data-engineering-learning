```python
def addition(num1, num2):
  return num1 + num2

def subtraction(num1, num2):
  return num1 - num2

def multiplication(num1, num2):
  return num1 * num2

def division(num1, num2):
  return num1 / num2
```

```python
print(addition(1, 2))
print(subtraction(1, 2))
print(multiplication(1, 2))
print(division(1, 2))
```
