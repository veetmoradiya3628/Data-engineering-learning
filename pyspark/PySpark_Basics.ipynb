```python
# Pandas dataframe vs. PySpark dataframe
# PySpark DAG - execute in particular most optimized order in stages

from pyspark.sql import SparkSession

```

```python
spark = SparkSession.builder.getOrCreate()
spark
```

```python
df = spark.read.parquet('/content/yellow_tripdata_2025-01.parquet')
```

```python
df.show(5)
```

```python
df.count()
```

```python
df.schema.names
```

```python
df.printSchema()
```

```python
df.describe(['passenger_count', 'total_amount']).show()
```

```python
df.select('passenger_count').show(5)
```

```python
from pyspark.sql.functions import col
df.select(col('passenger_count')).show(5)
```

```python
df.sort('total_amount', ascending=False).show(5)
```

```python
df.filter('Airport_fee > 0').show(5)
```

```python
df.filter('Airport_fee > 0 and passenger_count > 2').show(5)
```

```python
df.filter('Airport_fee > 0 or passenger_count > 2').select('VendorId', 'passenger_count', 'total_amount').show(5)
```

```python
# Write a query combining sort, select and filter that shows all non-airport rides with exactly 1 passenger

df.filter('Airport_fee = 0 and passenger_count = 1').select('trip_distance', 'total_amount').sort('trip_distance', ascending=False).show(5)
```
